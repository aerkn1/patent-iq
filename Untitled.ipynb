```python
from pathlib import Path

path = Path("etl/conf/build.yaml")
text = path.read_text(encoding="utf-8")

start = text.index("<<<<<<< Updated upstream")
mid = text.index("=======", start)
end = text.index(">>>>>>>", mid)

replacement = """  upload_connection_timeout_seconds: 60
upload_read_timeout_seconds: 1200
upload_retry_max_attempts: 3
upload_retry_backoff_seconds: 5
"""

text = text[:start] + replacement + text[end:].split("\n", 1)[1]
path.write_text(text, encoding="utf-8")
print("conflict block removed")
```
